```python
# Import the libraries
import numpy as np
import matplotlib.pyplot as plt

print("Ready to learn about machine learning!")
```

```python
# The number of hours each student studied
hours_studied = np.array([1, 2, 3, 4, 5, 6])

# Whether each student passed (0 = no, 1 = yes)
passed = np.array([0, 0, 0, 1, 1, 1])

# Display our data
print("Hours studied:", hours_studied)
print("Passed:", passed)
```

```python
# Create two starting values that the model will learn
weight = 0.0
bias = 0.0

# Calculate the model's prediction
def predict(hours):
    z = weight * hours + bias
    probability = 1 / (1 + np.exp(-z))
    return probability

# Make a prediction for a student who studied 5 hours
prediction = predict(5)

print("Predicted probability of passing:", prediction)
```

```python
# Start with weight and bias equal to zero
weight = 0.0
bias = 0.0

# Set how quickly the model learns
learning_rate = 0.1

# Train the model for 100 rounds
for round in range(100):

    # Make predictions
    z = weight * hours_studied + bias
    predictions = 1 / (1 + np.exp(-z))

    # Calculate how wrong the predictions are
    errors = predictions - passed

    # Calculate the gradients
    weight_gradient = np.mean(errors * hours_studied)
    bias_gradient = np.mean(errors)

    # Update the weight and bias
    weight = weight - learning_rate * weight_gradient
    bias = bias - learning_rate * bias_gradient

# Display the learned values
print("Learned weight:", weight)
print("Learned bias:", bias)

# Make predictions after training
final_predictions = 1 / (1 + np.exp(-(weight * hours_studied + bias)))

print("Final predictions:", final_predictions)
```

### Week 1 Notes: Logistic Regression and Gradient Descent

One important concept I learned was gradient descent. Gradients tell the model how to adjust its weights and bias to reduce prediction errors. Backpropagation uses this idea to calculate gradients in more complicated neural networks.

## 1. Binary Cross-Entropy Gradient Derivation

### What is cross-entropy loss?

Cross-entropy loss measures how wrong a model's prediction is compared to the actual answer.

- y = actual answer (0 or 1)
- p = predicted probability
- L = loss

The formula is:

L = -[y * ln(p) + (1-y) * ln(1-p)]

A smaller loss means the model made a better prediction. A larger loss means it made a worse prediction.

### What are we trying to find?

We want to figure out how the loss changes when we change z.

First, the model calculates:

z = weight * x + bias

Then it uses the sigmoid function to turn z into a probability:

p = 1 / (1 + e^(-z))

Finally, it calculates the loss using that probability.

### Step 1: Use the chain rule

Since the loss depends on p, and p depends on z, we use the chain rule to break the derivative into two parts.

dL/dz = (dL/dp) * (dp/dz)

Basically, we find each derivative separately and then multiply them together.

### Step 2: Differentiate the loss function

Starting with:

L = -[y * ln(p) + (1-y) * ln(1-p)]

We take the derivative with respect to p:

dL/dp = -y/p + (1-y)/(1-p)

### Step 3: Differentiate the sigmoid function

Our sigmoid function is:

p = 1 / (1 + e^(-z))

Its derivative is:

dp/dz = p * (1-p)

### Step 4: Multiply the two derivatives

Now we plug both derivatives into the chain rule:

dL/dz = [-y/p + (1-y)/(1-p)] * p * (1-p

```python
# Calculate binary cross-entropy loss
def calculate_loss(y, p):
    # Avoid taking the log of zero
    p = np.clip(p, 1e-12, 1 - 1e-12)

    loss = -(y * np.log(p) + (1 - y) * np.log(1 - p))
    return loss
```

```python
# Actual answer
y = 1

# Predicted probability
p = 0.9

# Calculate loss
loss = calculate_loss(y, p)

print("Loss:", loss)
```

```python
# Actual answer
y = 1

# Starting value
z = 2.0

# Calculate the prediction using sigmoid
p = 1 / (1 + np.exp(-z))

# Calculate the gradient using our formula
gradient_by_hand = p - y

print("Prediction:", p)
print("Gradient by hand:", gradient_by_hand)
```

```python
# Small change to z
epsilon = 0.000001

# Calculate the original loss
original_loss = calculate_loss(
    y,
    1 / (1 + np.exp(-z))
)

# Increase z slightly
new_z = z + epsilon

# Calculate the new loss
new_p = 1 / (1 + np.exp(-new_z))
new_loss = calculate_loss(y, new_p)

# Estimate the gradient
gradient_numerical = (new_loss - original_loss) / epsilon

print("Numerical gradient:", gradient_numerical)
```

```python
# Compare both gradients
print("Gradient by hand:", gradient_by_hand)
print("Numerical gradient:", gradient_numerical)

difference = abs(gradient_by_hand - gradient_numerical)

print("Difference:", difference)
```

### Numerical Gradient Check

I wanted to check whether the gradient I derived by hand matched the gradient calculated numerically using finite differences.

I used the same starting value for z and compared both methods.

**Results:**

- Gradient by hand: -0.119202922
- Numerical gradient: -0.119202870
- Difference: 0.0000000524

The difference was extremely small and below the required tolerance of 1e-6.

**Conclusion:**

The numerical gradient closely matched the gradient from my mathematical derivation. This means my gradient calculation passed the numerical gradient check.


```python
# Intentionally calculate the gradient incorrectly
wrong_gradient = y - p

# Compare it with the numerical gradient
print("Wrong gradient:", wrong_gradient)
print("Numerical gradient:", gradient_numerical)

# Calculate the difference
wrong_difference = abs(wrong_gradient - gradient_numerical)

print("Difference:", wrong_difference)

# Check whether the gradient passes
if wrong_difference < 1e-6:
    print("Gradient check passed")
else:
    print("Gradient check failed")
```
